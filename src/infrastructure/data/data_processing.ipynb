```python
import pandas as pd
data = pd.read_csv("matches.csv")
data.head()
```

```python
data.isnull().sum()
```

```python
filtered_data = data[data['venue'] == "Home"]
filtered_data['start_time'] = filtered_data['start_time'].str.replace(r'\s*\(.*?\)', '', regex=True)
filtered_data.to_csv("filtered_matches.csv", index=False, header=True)

```
